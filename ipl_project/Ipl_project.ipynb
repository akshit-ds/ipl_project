# Here's some basic questions

## 1.Which team won the most matches

```python
highestt_match_winner = df['match_winner'].value_counts()
```

```python
highestt_match_winner
```

```python
sns.barplot(y=highestt_match_winner.index,x=highestt_match_winner.values)
```

## 2.Toss decision trends

```python
df['toss_decision'].value_counts()
```

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title('Toss trends')
```

## Toss decisions vs Winner

```python
df[df['toss_winner']==df['match_winner']]['toss_winner'].count()
```

```python
per = df[df['toss_winner']==df['match_winner']]['toss_winner'].count()
percentage = (per*100)/df.shape[0]
percentage.round(2)
```
